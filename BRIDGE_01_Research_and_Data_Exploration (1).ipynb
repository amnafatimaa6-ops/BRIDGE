```python
BASE = Path("/content/PROJECT_BRIDGE")

experiment_folder = BASE / "04_Experiments"
experiment_folder.mkdir(parents=True, exist_ok=True)

#  Research protocol
protocol = """
# PROJECT BRIDGE — Experimental Protocol

## Research objective
Compare human-only and AI-assisted workflows to investigate
whether AI creates measurable operational value for NOVA Retail.

## Experiment A: Customer Enquiry Management
Benchmark: 100 synthetic customer enquiries.

Human-only:
- Read the enquiry and fictional policy.
- Write the response independently.
- Record completion time.

AI-assisted:
- Provide the same enquiry and policy to the AI system.
- Record AI processing time.
- Review and correct the generated response.
- Record total elapsed time and human review time.

Evaluation:
- Policy accuracy.
- Completeness.
- Unsupported claims.
- Escalation correctness.
- Correction count.
- Time and estimated labour cost.

## Experiment B: Customer Feedback Analysis
Benchmark: 120 synthetic customer reviews.

Tasks:
- Sentiment classification.
- Feedback topic classification.

Model input:
- review_text only.

Targets:
- sentiment_label.
- feedback_topic.

Evaluation:
- Accuracy.
- Macro precision.
- Macro recall.
- Macro F1.
- Confusion matrix.
- Processing time.

Use a stratified train/test split.
Keep the test set separate from model training and model selection.

Important limitation:
The labels were manually constructed by one annotator.
Results measure agreement with this synthetic annotation scheme,
not independently verified real-world sentiment.

Do not claim an independent human accuracy comparison unless
a separate reviewer independently labels the test records.

## Experiment C: Product Description Generation
Benchmark: 30 synthetic product records.

Human-only:
- Write a product description using catalogue facts.

AI-assisted:
- Generate a description using the same catalogue facts.
- Review and correct the AI draft.

Evaluation:
- Factual accuracy.
- Completeness.
- Readability.
- Consistency.
- Unsupported claims.
- Human correction time.
- Total completion time.

## Fairness and consistency
- Both workflows receive the same source information.
- Use a fixed task brief and evaluation rubric.
- Record the AI model and version where available.
- Record prompts and model settings.
- Record AI processing time separately from human review time.
- Randomise or alternate task order where practical.
- Do not change evaluation criteria after seeing results.

## Financial evaluation
Calculate labour cost using measured human time and an explicitly
stated hourly labour-rate assumption.

Include AI subscription/API costs and estimated error-related costs
where reliable evidence exists.

Separate measured results from assumptions and projections.

## Research limitations
All three datasets are synthetic.
The experiments are controlled pilot tests, not evidence of
performance in an actual operating business.
"""

protocol_path = experiment_folder / "experimental_protocol.md"
protocol_path.write_text(protocol.strip(), encoding="utf-8")

#   Experiment A logging template
enquiry_log = pd.DataFrame(columns=[
    "task_id",
    "human_time_seconds",
    "ai_processing_seconds",
    "ai_review_seconds",
    "ai_total_time_seconds",
    "human_quality_score",
    "ai_quality_score",
    "ai_correction_count",
    "ai_unsupported_claims",
    "ai_escalation_correct",
    "notes"
])

#    Experiment B logging template
feedback_log = pd.DataFrame(columns=[
    "review_id",
    "true_sentiment",
    "predicted_sentiment",
    "true_topic",
    "predicted_topic",
    "processing_time_seconds"
])

#   Experimenting C logging template
product_log = pd.DataFrame(columns=[
    "product_id",
    "human_time_seconds",
    "ai_processing_seconds",
    "ai_review_seconds",
    "ai_total_time_seconds",
    "human_quality_score",
    "ai_quality_score",
    "ai_correction_count",
    "ai_unsupported_claims",
    "notes"
])

logs = {
    "experiment_A_enquiries_log.csv": enquiry_log,
    "experiment_B_feedback_log.csv": feedback_log,
    "experiment_C_products_log.csv": product_log
}

for filename, dataframe in logs.items():
    dataframe.to_csv(experiment_folder / filename, index=False)

print("EXPERIMENT WORKSPACE CREATED")
print("\nFiles:")

for path in sorted(experiment_folder.iterdir()):
    print(path.name)

print("\nExperiment protocol saved.")
print("Three experiment logging templates saved.")
```

```python
# Project directory
BASE = Path("/content/PROJECT_BRIDGE")
EXP = BASE / "04_Experiments"
EXP.mkdir(parents=True, exist_ok=True)

# five uploaded CSV files
batch_files = [
    Path("/content/enquiry_responses.csv"),
    Path("/content/enquiry_responses_2.csv"),
    Path("/content/enquiry_responses_3.csv"),
    Path("/content/enquiry_responses_4.csv"),
    Path("/content/enquiry_responses_5.csv")
]

# columns
expected_columns = [
    "task_id",
    "enquiry_text",
    "category",
    "complexity",
    "human_response",
    "human_time_seconds",
    "notes"
]

# Check that all files exist
for file in batch_files:
    assert file.exists(), f"Missing file: {file}"

# Load all batches
batches = []

for i, file in enumerate(batch_files, start=1):
    df = pd.read_csv(file, keep_default_na=False)

    missing_columns = set(expected_columns) - set(df.columns)
    assert not missing_columns, (
        f"Batch {i} is missing columns: {missing_columns}"
    )

    df = df[expected_columns].copy()
    df["batch"] = i

    batches.append(df)

    print(f"Batch {i}: {len(df)} records loaded")

# Merging batches
human_baseline = pd.concat(batches, ignore_index=True)

# Converting numerical fields
human_baseline["task_id"] = pd.to_numeric(
    human_baseline["task_id"], errors="coerce"
)

human_baseline["human_time_seconds"] = pd.to_numeric(
    human_baseline["human_time_seconds"], errors="coerce"
)

# Sorting by original task ID
human_baseline = human_baseline.sort_values(
    "task_id"
).reset_index(drop=True)

# Save as a new master dataset
output_path = EXP / "experiment_A_human_baseline_final.csv"

# Structural validation
assert len(human_baseline) == 100, "Expected 100 records"
assert human_baseline["task_id"].tolist() == list(range(1, 101)), (
    "Task IDs are missing, duplicated or out of order"
)

human_baseline.to_csv(output_path, index=False)

print("\nHUMAN BASELINE MERGE COMPLETE")
print("Total records:", len(human_baseline))
print("Unique task IDs:", human_baseline["task_id"].nunique())
print("Saved to:", output_path)

display(human_baseline.head())
```

```python
# Checking missing values
missing_values = human_baseline[
    expected_columns
].isna().sum()

# Checking empty text fields
empty_responses = (
    human_baseline["human_response"]
    .astype(str).str.strip().eq("")
).sum()

empty_enquiries = (
    human_baseline["enquiry_text"]
    .astype(str).str.strip().eq("")
).sum()

# Checking timing validity
invalid_times = (
    human_baseline["human_time_seconds"].isna()
    | (human_baseline["human_time_seconds"] <= 0)
).sum()

# Checking duplicate task IDs
duplicate_ids = human_baseline["task_id"].duplicated().sum()

# Overall timing
total_time = human_baseline["human_time_seconds"].sum()
average_time = human_baseline["human_time_seconds"].mean()
median_time = human_baseline["human_time_seconds"].median()

print("DATA QUALITY AUDIT")
print("-------------------")
print("Missing values:\n", missing_values)
print("Empty responses:", empty_responses)
print("Empty enquiries:", empty_enquiries)
print("Invalid timings:", invalid_times)
print("Duplicate task IDs:", duplicate_ids)

print("\nHUMAN BASELINE TIMING")
print("--------------------")
print(f"Total time: {total_time:.0f} seconds")
print(f"Total time: {total_time / 60:.2f} minutes")
print(f"Average response time: {average_time:.2f} seconds")
print(f"Median response time: {median_time:.2f} seconds")

print("\nTIMING BY COMPLEXITY")

complexity_summary = (
    human_baseline.groupby("complexity")["human_time_seconds"]
    .agg(["count", "mean", "median", "sum"])
    .round(2)
)

display(complexity_summary)

print("\nTIMING BY CATEGORY")

category_summary = (
    human_baseline.groupby("category")["human_time_seconds"]
    .agg(["count", "mean", "median", "sum"])
    .round(2)
)

display(category_summary)
```

```python
import json
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Project paths
DATA_PATH = Path(
    "/content/PROJECT_BRIDGE/data/processed/customer_feedback_final.csv"
)

EXP_B = Path("/content/PROJECT_BRIDGE/04_Experiments")
EXP_B.mkdir(parents=True, exist_ok=True)

# Loading dataset
feedback = pd.read_csv(DATA_PATH)

# Basic validation
required_columns = [
    "review_id",
    "review_text",
    "sentiment_label",
    "feedback_topic"
]

assert all(col in feedback.columns for col in required_columns)
assert len(feedback) == 120
assert feedback[required_columns].notna().all().all()
assert feedback["review_id"].is_unique

print("DATASET LOADED")
print("Shape:", feedback.shape)
print("\nSentiment distribution:")
display(feedback["sentiment_label"].value_counts())

print("\nTopic distribution:")
display(feedback["feedback_topic"].value_counts())

# Prepare text
X = feedback["review_text"].astype(str)

targets = {
    "sentiment": feedback["sentiment_label"],
    "topic": feedback["feedback_topic"]
}

models = {}
results = []
test_records = {}

for task_name, y in targets.items():

    # Independent stratified split for each prediction task
    X_train, X_test, y_train, y_test, id_train, id_test = (
        train_test_split(
            X,
            y,
            feedback["review_id"],
            test_size=0.20,
            random_state=42,
            stratify=y
        )
    )

    # TF-IDF + Logistic Regression pipeline
    model = Pipeline([
        ("tfidf", TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2),
            max_features=5000,
            min_df=1
        )),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ])

    # Train
    model.fit(X_train, y_train)

    # Predict held-out test records
    predictions = model.predict(X_test)

    # Calculate metrics
    metrics = {
        "task": task_name,
        "accuracy": accuracy_score(y_test, predictions),
        "macro_precision": precision_score(
            y_test, predictions, average="macro",
            zero_division=0
        ),
        "macro_recall": recall_score(
            y_test, predictions, average="macro",
            zero_division=0
        ),
        "macro_f1": f1_score(
            y_test, predictions, average="macro",
            zero_division=0
        )
    }

    results.append(metrics)
    models[task_name] = model

    test_records[task_name] = {
        "train_ids": id_train.tolist(),
        "test_ids": id_test.tolist(),
        "actual": y_test.tolist(),
        "predicted": predictions.tolist()
    }

    print("\n" + "=" * 55)
    print(task_name.upper(), "CLASSIFICATION")
    print("=" * 55)

    print(classification_report(
        y_test,
        predictions,
        zero_division=0
    ))

    print("Confusion matrix:")
    labels = sorted(y.unique())

    cm = confusion_matrix(
        y_test,
        predictions,
        labels=labels
    )

    display(pd.DataFrame(
        cm,
        index=[f"Actual: {x}" for x in labels],
        columns=[f"Predicted: {x}" for x in labels]
    ))

    print("Test records:", len(y_test))
    print("Training records:", len(y_train))

results_df = pd.DataFrame(results)

results_df.to_csv(
    EXP_B / "experiment_B_model_metrics.csv",
    index=False
)

with open(
    EXP_B / "experiment_B_test_predictions.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(test_records, f, indent=4)

print("\n" + "=" * 55)
print("EXPERIMENT B BASELINE RESULTS")
display(results_df.round(4))

print("Metrics and test predictions saved.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)


svm_models = {}
svm_results = []

for task_name, target_col in [
    ("sentiment", "sentiment_label"),
    ("topic", "feedback_topic")
]:

    records = test_records[task_name]

    train_data = feedback[
        feedback["review_id"].isin(records["train_ids"])
    ]

    test_data = feedback[
        feedback["review_id"].isin(records["test_ids"])
    ]

    X_train = train_data["review_text"]
    y_train = train_data[target_col]

    X_test = test_data["review_text"]
    y_test = test_data[target_col]

    svm = Pipeline([
        ("tfidf", TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2),
            max_features=5000,
            min_df=1
        )),
        ("classifier", LinearSVC(random_state=42))
    ])

    svm.fit(X_train, y_train)

    predictions = svm.predict(X_test)

    svm_models[task_name] = svm

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    svm_results.append({
        "task": task_name,
        "model": "Linear SVM",
        "accuracy": accuracy_score(y_test, predictions),
        "macro_precision": precision,
        "macro_recall": recall,
        "macro_f1": f1
    })

svm_results_df = pd.DataFrame(svm_results)

display(svm_results_df.round(3))
```

```python
# Actual measured human baseline
human_seconds = comparison_df["human_time_seconds"].mean()
human_total_seconds = comparison_df["human_time_seconds"].sum()

print("MEASURED HUMAN BASELINE")
print("Tasks:", len(comparison_df))
print("Average time per task:", round(human_seconds, 2), "seconds")
print("Total time:", round(human_total_seconds / 60, 2), "minutes")

# Illustrative assumptions — not measured business costs
hourly_wages = [200, 300, 500]       # PKR per hour
ai_costs = [0, 1, 5]                 # PKR per enquiry

rows = []

for wage in hourly_wages:
    human_cost = (human_seconds / 3600) * wage

    for ai_cost in ai_costs:
        max_review_seconds = (
            (human_cost - ai_cost) / wage * 3600
        )

        rows.append({
            "assumed_hourly_wage_pkr": wage,
            "assumed_ai_cost_per_task_pkr": ai_cost,
            "human_only_cost_per_task_pkr": round(human_cost, 3),
            "maximum_break_even_review_seconds": round(
                max(0, max_review_seconds), 2
            )
        })

break_even_df = pd.DataFrame(rows)

display(break_even_df)

break_even_df.to_csv(
    EXP_B / "experiment_A_break_even_scenarios.csv",
    index=False
)

print("Break-even scenarios saved.")
```
